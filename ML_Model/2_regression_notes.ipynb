## What is regression ?
Regression is a way to predict a number.
It helps us find out how one thing changes when something else changes.

Imagine you're running a lemonade stand. You've been keeping track of:
- How many glasses you sell each day
- And what the temperature was on that day
- After a while, you notice:
<br> **“The hotter the day, the more lemonade I sell."**<br>
Regression is the math that helps you figure out exactly how much more lemonade you'll sell as the temperature goes up.

**Regression is about finding patterns in number so that we can make smart predictions.**

## Linear Regression

Linear regression is a supervised machine learning algorithm that models the linear relationship between a dependent target variable and one or more independent features by fitting a straight line (or hyperplane) that minimizes errors.

It is primarily used for predicting continuous numeric values, such as forecasting prices or trends.

### Key Aspects of Linear Regression
- Equation: The formula is y = mx + c, where 
y is the prediction, 
x is the input, 
m is the slope (weight), and 
c is the intercept.

- Goal: To find the "best-fit" line by minimizing the sum of squared errors (SSE) between predicted and actual data points.

### Overfitting and Underfitting

Case 1: Overfitting
- Training: model perform better (low bias)
- Testing: model not performed good (high variance)

Case 2: Underfitting
- Training: model not performed good (high bias)
- Testing: model not performed good (high variance)

Perfect Model:
- Low bias
- Low variance
