## KNN (Kth Nearest Neigbhour)

The K-Nearest Neighbors (KNN) algorithm is a simple, supervised machine learning algorithm used for both classification and regression tasks.

### Working of KNN algorithm

- Step 1: Selecting the optimal value of K
- Step 2: Calculating distance
- Step 3: Finding Nearest Neighbors
- Step 4: Voting for Classification or Taking Average for Regression

### Implementation of KNN

Loading the data

```python
from sklearn.datasets import make_moons
import pandas as pd

# Create synthetic 2D data
X, y = make_moons(n_samples=300, noise=0.3, random_state=42)

# Create a DataFrame for plotting
df = pd.DataFrame(X, columns=["Feature 1", "Feature 2"])
df['Target'] = y
```

Train - Test Split &  Normalization

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Normalize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split into train and test
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42, stratify=y
)
```

Fit the k-NN Model and Evaluate

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Train a k-NN classifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

# Predict and evaluate
y_pred = knn.predict(X_test)
print(f"Test Accuracy (k=5): {accuracy_score(y_test, y_pred):.2f}")
```
