# Model Tuning

Model tuning (or hyperparameter tuning) is the process of finding the best settings (hyperparameters) for a machine learning model to improve its performance.

👉 These settings are not learned from data, but are chosen before training.

### Intitution

Think of a model like a car engine ⚙️:

- The model = engine
- Hyperparameters = knobs (speed, fuel mix, pressure)

👉 If you don’t adjust them properly:

- Too slow → underperformance
- Too aggressive → crashes (overfitting)

# Cross-Validation

### The problem
When training an ML model, the standard approach is to split your data into two parts: a Training Set (to learn from) and a Testing Set (to evaluate performance). 

But what if the data in your testing set is accidentally entirely easy or entirely hard? Your model's performance score would be misleading.

### The Solution: K-Fold Cross-Validation
Instead of splitting the data just once, K-Fold Cross Validation splits the data multiple times to ensure every single data point gets to be in the "test" group exactly once.

Here is how the standard K-Fold process works:

1. Shuffle the dataset randomly.

2. Split the dataset into K equal-sized groups (called "folds"). K is typically 5 or 10.

3. Iterate: For each of the K folds:

- Keep that single fold aside as the Validation/Test set.

- Take the remaining K-1 folds, combine them, and use them as the Training set.

- Train your model and record its accuracy/score.

4. Average: Once you have done this K times, you will have K different accuracy scores. You average them together to get your final, highly reliable performance metric.

### Why do we use it?
- More Reliable Scoring: It gives a much more accurate picture of how your model will perform on real-world, unseen data.

- Maximum Data Usage: Every single data point gets used for both training and testing, which is incredibly valuable when your dataset is small.
