### Summary: Traditional Programming vs Machine Learning

| Comparison | Traditional Programming          | Machine Learning                 |
| ---------- | -------------------------------- | -------------------------------- |
| **Input**  | Rules (if–else logic)            | Data (examples)                  |
| **Output** | Executes hard-coded rules        | Learns patterns & predicts       |
| **Logic**  | Explicit logic written by humans | Implicit logic learned from data |

---

### Difference Between AI, ML, and DL

Artificial Intelligence (AI):
Broad field where machines act intelligently.

Machine Learning (ML):
Subset of AI where machines learn from data.

Deep Learning (DL):
Subset of ML using neural networks with many layers.

Hierarchy:
AI → ML → DL

### Definitions

**Classification**
Predicting discrete categories
*Examples:* PASS/FAIL, Spam/Not Spam, Cat/Dog

**Regression**
Predicting continuous numbers
*Examples:* Marks, Price, Temperature

### Types of Learning

2. Unsupervised Learning

4. Reinforcement Learning
---


### Supervised Learning

> Data has **inputs + correct outputs**

**Definition**

```
SUPERVISED LEARNING
Learning from labeled data (inputs + outputs)

Includes:
• Classification
• Regression
```

---
### Unsupervised Learning

> Data has **inputs only**, no correct answers.

**Definition**

```
UNSUPERVISED LEARNING
Finding hidden structures without labels

Examples:
• Customer Segmentation
• Clustering
• Grouping Images
```

---

### Semi-Supervised Learning

Small labeled data + large unlabeled data

```
Example:
Doctor labels few X-rays
Model labels the rest
```

---

### Reinforcement Learning

* Agent
* Environment
* Action
* Reward
* Penalty

> Learning through **trial and error**

| Action    | Result      |
| --------- | ----------- |
| Good move | Reward (+)  |
| Bad move  | Penalty (–) |

**Key Insight:**
The system generates its **own data**.

### EDA
EDA stands for Exploratory Data Analysis - it's the step where you explore the data to:
- Understand it
- Discover patterns
- Spot anomalies
- Generate insights
- And decide what to do next

### EDA Steps:

**1. Viewing the data**
- head(), tail(), shape, info()
- What columns do i have?
- What types of data?

**2. Summary statistics**
- mean, median, mode, std, min, max, quartiles
- Helps understand spread and central tendency

**3. Value Counts**
- How many unique values in a column?
- Great for categorical columns

**4. Missing Value Analysis**
- Where are the gaps?
- What percent of the data is missing?

**5. Visualizations**
- Histograms → distribution of values
- Boxplots → outliers and spread
- Bar plots → comparisons of categories
- Correlation heatmaps → linear relationships between numerical features
- Scatter plots → bivariate relationships

**6. Target Variable Exploration**
- How does your output (like 'charges' in your dataset) relate to other variables?

### EDA Importance
- You can't clean or preprocess what you don't understand.
- It helps you identify mistakes, biases, or limitations in the data.
- It guides the direction of your data cleaning and feature engineering.
- It gives your audience or stakeholders a “story” or overview of what the data is telling you.
