### Summary: Traditional Programming vs Machine Learning

| Comparison | Traditional Programming          | Machine Learning                 |
| ---------- | -------------------------------- | -------------------------------- |
| **Input**  | Rules (if–else logic)            | Data (examples)                  |
| **Output** | Executes hard-coded rules        | Learns patterns & predicts       |
| **Logic**  | Explicit logic written by humans | Implicit logic learned from data |

---

### Difference Between AI, ML, and DL

Artificial Intelligence (AI):
Broad field where machines act intelligently.

Machine Learning (ML):
Subset of AI where machines learn from data.

Deep Learning (DL):
Subset of ML using neural networks with many layers.

Hierarchy:
AI → ML → DL

---

### Definitions

**Classification**
Predicting discrete categories
*Examples:* PASS/FAIL, Spam/Not Spam, Cat/Dog

**Regression**
Predicting continuous numbers
*Examples:* Marks, Price, Temperature

---

### Types of Learning

1. Supervised Learning

2. Unsupervised Learning

3. Semi-Supervised Learning

4. Reinforcement Learning
---


### Supervised Learning

> Data has **inputs + correct outputs**

**Definition**

```
SUPERVISED LEARNING
Learning from labeled data (inputs + outputs)

Includes:
• Classification
• Regression
```

---
### Unsupervised Learning

> Data has **inputs only**, no correct answers.

**Definition**

```
UNSUPERVISED LEARNING
Finding hidden structures without labels

Examples:
• Customer Segmentation
• Clustering
• Grouping Images
```

---

### Semi-Supervised Learning

Small labeled data + large unlabeled data

```
Example:
Doctor labels few X-rays
Model labels the rest
```

---

### Reinforcement Learning

* Agent
* Environment
* Action
* Reward
* Penalty

> Learning through **trial and error**

| Action    | Result      |
| --------- | ----------- |
| Good move | Reward (+)  |
| Bad move  | Penalty (–) |

**Key Insight:**
The system generates its **own data**.
